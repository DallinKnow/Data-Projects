## Import Libraries 📥

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

## Import and Read Data 📥📚

```python
file_groups = {
    "daily": ["dailyActivity_merged.csv", "dailyCalories_merged.csv", "dailyIntensities_merged.csv", "dailySteps_merged.csv"],
    "hourly": ["hourlyCalories_merged.csv", "hourlyIntensities_merged.csv", "hourlySteps_merged.csv"],
    "minute": ["minuteCaloriesNarrow_merged.csv", "minuteCaloriesWide_merged.csv", "minuteIntensitiesNarrow_merged.csv", "minuteIntensitiesWide_merged.csv", "minuteMETsNarrow_merged.csv", "minuteSleep_merged.csv", "minuteStepsNarrow_merged.csv", "minuteStepsWide_merged.csv"],
    "info": ["heartrate_seconds_merged.csv", "sleepDay_merged.csv", "weightLogInfo_merged.csv"]
}

data = {}
for group, files in file_groups.items():
    data[group] = {
    f.replace("_merged.csv", ""): pd.read_csv(os.path.join('data/mturkfitbit_export_4.12.16-5.12.16/Fitabase data 4.12.16-5.12.16', f))
    for f in files
    }
```

## Checking Data Frames 👀

```python
for group, frame in data.items() :
    print(f"\n========{group.upper()}===========")
    for name, df in frame.items(): 
        print(f"\n------{name}-------")
        display(df.head())
```

## Cleaning 🧼🧹

```python
# Running many functions to comb through and check for any dirty data.

for group, frame in data.items() :
    print(f"\n========{group.upper()}===========")
    for name, df in frame.items(): 
        print(f"\n------{name}-------")
        display(df.isnull().sum()) # Here I am lookng for null values
        print("\n Duplicates: ")
        display(df.duplicated().sum()) # Here I am looking for duplicates
        print("\n Info: ")
        display(df.describe()) # Here I am checking for outlires and misinput values (ex. -1)
```
