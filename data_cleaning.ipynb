```python
# Importing pandas
import pandas as pd
```

```python
# Loading 2025 monthly data into separate DataFrames to be merged later
df_1 = pd.read_csv('202501-divvy-tripdata.csv')
df_2 = pd.read_csv('202502-divvy-tripdata.csv')
df_3 = pd.read_csv('202503-divvy-tripdata.csv')
df_4 = pd.read_csv('202504-divvy-tripdata.csv')
df_5 = pd.read_csv('202505-divvy-tripdata.csv')
df_6 = pd.read_csv('202506-divvy-tripdata.csv')
df_7 = pd.read_csv('202507-divvy-tripdata.csv')
df_8 = pd.read_csv('202508-divvy-tripdata.csv')
df_9 = pd.read_csv('202509-divvy-tripdata.csv')
df_10 = pd.read_csv('202510-divvy-tripdata.csv')
df_11 = pd.read_csv('202511-divvy-tripdata.csv')
df_12 = pd.read_csv('202512-divvy-tripdata.csv')
# Loading 2024 monthly data into separate DataFrames to be merged later
df_13 = pd.read_csv('202401-divvy-tripdata.csv')
df_14 = pd.read_csv('202402-divvy-tripdata.csv')
df_15 = pd.read_csv('202403-divvy-tripdata.csv')
df_16 = pd.read_csv('202404-divvy-tripdata.csv')
df_17 = pd.read_csv('202405-divvy-tripdata.csv')
df_18 = pd.read_csv('202406-divvy-tripdata.csv')
df_19 = pd.read_csv('202407-divvy-tripdata.csv')
df_20 = pd.read_csv('202408-divvy-tripdata.csv')
df_21 = pd.read_csv('202409-divvy-tripdata.csv')
df_22 = pd.read_csv('202410-divvy-tripdata.csv')
df_23 = pd.read_csv('202411-divvy-tripdata.csv')
df_24 = pd.read_csv('202412-divvy-tripdata.csv')
```

```python
# Checking the dataframes for consistency before merging
df_1.info()
df_2.info()
df_3.info()
df_4.info()
df_5.info()
df_6.info()
df_7.info()
df_8.info()
df_9.info()
df_10.info()
df_11.info()
df_12.info()
df_13.info()
df_14.info()
df_15.info()
df_16.info()
df_17.info()
df_18.info()
df_19.info()
df_20.info()
df_21.info()
df_22.info()
df_23.info()
df_24.info()
```

```python
# Combine all monthly DataFrames into one for the entire 2024-2025 period
df = pd.concat([df_1, df_2, df_3, df_4, df_5, df_6, df_7, df_8, df_9, df_10, df_11, df_12, df_13, df_14, df_15, df_16, df_17, df_18, df_19, df_20, df_21, df_22, df_23, df_24], ignore_index=True)
```

```python
# Checking the combined DataFrame after merging
df.head()
```

```python
# Checking for null values in the combined DataFrame
df.isnull().sum()
```

```python
# Found null values, but lattitude and longitude end values are not equal to start station and end station values
# Dropping null values in end_latitude and end_longitude columns
df.dropna(subset=['end_lat', 'end_lng'], inplace=True)
```

```python
# Checking for null values in the combined DataFrame after dropping
df.isnull().sum()
```

```python
# End station nulls not equal to start station for some entries - dropping all null values for simplicity
df.dropna(inplace=True)
```

```python
# Checking for null values in the combined DataFrame after dropping
df.isnull().sum()
```

```python
# All null values have been addressed
# Checking for duplicate entries in the combined DataFrame
df.duplicated().sum()
```

```python
# No duplicate entries found
# Checking the values after cleaning
df['member_casual'].value_counts()
```

```python
# Checking the values after cleaning
df['rideable_type'].value_counts()
```

```python
# Checking values for consistency in start and end time columns to measure ride duration
df['started_at'].head()
df['ended_at'].head()
```

```python
# Creating a new column for ride duration by finding the difference between end time and start time
df['ride_duration'] = pd.to_datetime(df['ended_at']) - pd.to_datetime(df['started_at']) 
```

```python
# Found inconsistant date time stamps, finding the number of invalid date time stamps in start and end time columns
started_temp = pd.to_datetime(df['started_at'], errors='coerce')
invalid_started = started_temp.isna().sum()
print(f"Invalid 'started_at' count: {invalid_started}")
```

```python
# Found inconsistant date time stamps,, finding the number of invalid date time stamps in start and end time columns
ended_temp = pd.to_datetime(df['ended_at'], errors='coerce')
invalid_ended = ended_temp.isna().sum()
print(f"Invalid 'ended_at' count: {invalid_ended}")
```

```python
# The count on both start and end time columns are the same, so we can reformat the invalid date time stamps to NaT
# and recalculate the ride duration
df['ride_duration'] = pd.to_datetime(df['ended_at'], errors='coerce') - pd.to_datetime(df['started_at'], errors='coerce')
```

```python
# Checking for null values in the combined DataFrame after reformatting date time stamps
df.isna().sum()
```

```python
# Dropping null values from ride duration
df.dropna(inplace=True)
```

```python
# Checking data after dropping null values from ride duration
df.isna().sum()
```

```python
# Looking for out of the ordinary values in ride duration
df['ride_duration'].describe()
```

```python
# finding how many rides have a duration of 10 seconds or less for false entries in the data or nonriders
zero_or_negative_duration = (df['ride_duration'] <= pd.Timedelta(seconds=10)).sum()
print(f"Number of rides with duration of 10 seconds or less: {zero_or_negative_duration}")
```

```python
# finding how many rides have a duration of 24 hours or more for simplicity in the data and to drop outliers that will skew the data
long_duration = (df['ride_duration'] >= pd.Timedelta(24, unit='hours')).sum()
print(f"Number of rides with duration of 24 hours or more: {long_duration}")
```

```python
# Dropping rides with duration of 10 seconds or less or more than 24 hours
df = df[(df['ride_duration'] > pd.Timedelta(seconds=10)) & (df['ride_duration'] < pd.Timedelta(24, unit='hours'))] 
```

```python
# Looking for out of the ordinary values in ride duration after dropping outliers
df['ride_duration'].describe()
```

```python
# Checking for null or bad syntax in started_at column
started_temp = pd.to_datetime(df['started_at'], errors='coerce')
invalid_started = started_temp.isna().sum()
print(f"Invalid 'started_at' count: {invalid_started}")
```

```python
# Checking for null
df.isnull().sum()
```

```python
# With data cleaned, I will now prepare three last columns for analysis: day of the week, month, and hour of the day. 
# I will extract this information from the start time column.
df['started_at'] = pd.to_datetime(df['started_at'])
df['start_day'] = df['started_at'].dt.day_name()
df['start_month'] = df['started_at'].dt.month_name()
df['start_hour'] = df['started_at'].dt.hour
```
